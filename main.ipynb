```python
target_column = "Target_Variable/Total Income"
feature_columns = [
    column
    for column in df.select_dtypes(include="number").columns
    if column != target_column
]

X = df[feature_columns].values
y = df[target_column].values

```

```python
import numpy as np

class Perceptron:
    def __init__(self, learning_rate=0.01, epochs=100):
        self.lr = learning_rate
        self.epochs = epochs
        self.weights = None
        self.bias = None
        
    def _step_function(self, x):
        return np.where(x >= 0, 1, 0)
        
    def fit(self, X, y):
        num_samples, num_features = X.shape
        
        self.weights = np.random.uniform(-0.5, 0.5, num_features)
        self.bias = np.random.uniform(-0.5, 0.5)
        
        for epoch in range(self.epochs):
                linear_output = np.dot(X, self.weights) + self.bias
                y_predicted = self._step_function(linear_output)
                update = self.lr * (y - y_predicted)
                
                self.weights += update * X.T
                self.bias += update
                
    def predict(self, X):
        linear_output = np.dot(X, self.weights) + self.bias
        return self._step_function(linear_output)
```

```python
import random 
import numpy as np

x1=np.array([1, 1, 0, 0])
x2=np.array([1, 0, 1, 0])
y=np.array([1, 1, 1, 0])
def array_to_individual(x1, x2, y):
    for i in range(len(x1)):
        yield x1[i], x2[i], y[i]

def step_function(x):
    return 1 if x >= 0 else 0
def perceptron(x1, x2, y, learning_rate=0.1, epochs=100):
    weights = [random.uniform(-0.5, 0.5), random.uniform(-0.5, 0.5)]
    bias = random.uniform(-0.5, 0.5)
    
    for epoch in range(epochs):
        linear_output = weights[0] * x1 + weights[1] * x2 + bias
        y_predicted = step_function(linear_output)
        
        update = learning_rate * (y - y_predicted)
        weights[0] += update * x1
        weights[1] += update * x2
        bias += update
        
    return weights, bias, y_predicted
perceptron(x1, x2, y)
```
