```python
import numpy as np

class Perceptron:
    def __init__(self, learning_rate=0.01, epochs=100):
        self.lr = learning_rate
        self.epochs = epochs
        self.weights = None
        self.bias = None

    def _step_function(self, x):
        return np.where(x >= 0, 1, 0)

    def fit(self, X, y):
        num_samples, num_features = X.shape

        self.weights = np.random.uniform(-0.5, 0.5, num_features)
        self.bias = np.random.uniform(-0.5, 0.5)

        for epoch in range(self.epochs):
            for sample_index in range(num_samples):
                linear_output = np.dot(X[sample_index], self.weights) + self.bias
                y_predicted = self._step_function(linear_output)
                update = self.lr * (y[sample_index] - y_predicted)

                self.weights += update * X[sample_index]
                self.bias += update

    def predict(self, X):
        linear_output = np.dot(X, self.weights) + self.bias
        return self._step_function(linear_output)
```

```python
# Example of AND Gate

X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])
y = np.array([0, 0, 0, 1])
```

```python
P = Perceptron(learning_rate = 0.1, epochs = 10)
P.fit(X, y)
```

```python
P.predict(X)
```
